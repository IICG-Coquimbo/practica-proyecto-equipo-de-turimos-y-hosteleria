# Acto 1: El Storytelling del EDA (El diagnóstico inicial) 

## Código para el Nivel 1: Estratégico

```python
from pymongo import MongoClient
import pandas as pd


client = MongoClient(
    "mongodb+srv://lucascheuque_db_user:27032005@cluster0.tjvu2a3.mongodb.net/"
)


data = list(
    client.proyecto_bigdata.alojamientos.find({}, {"_id":0})
)


df_completos = spark.createDataFrame(
    pd.DataFrame(data)
)


total_alojamientos = df_completos.count()


kpi_estrategico = df_completos.groupBy("plataforma") \
    .agg(
        F.count("plataforma").alias("Cantidad_Alojamientos"),

        F.round(
            (F.count("plataforma") / total_alojamientos)*100,
            2
        ).alias("Participacion_Porcentaje"),

        F.round(
            F.avg(F.col("precio_noche").cast("double")),
            0
        ).alias("Precio_Promedio_Noche")
    ) \
    .orderBy(
        F.desc("Participacion_Porcentaje")
    )


print(
"[KPI ESTRATEGICO] PARTICIPACION POR PLATAFORMA"
)

kpi_estrategico.show()
```

```python
import matplotlib.pyplot as plt


df_est_pandas = kpi_estrategico.toPandas()


plt.figure(figsize=(10,5))


plt.bar(
    df_est_pandas["plataforma"],
    df_est_pandas["Participacion_Porcentaje"]
)


plt.title(
    "Participacion de Alojamientos por Plataforma"
)


plt.xlabel("Plataforma")
plt.ylabel("% Participacion")


plt.xticks(rotation=45)

plt.show()
```

## Código para el Nivel 2: Táctico

```python
# KPI Tactico: bandas de precios por zona geografica

df_tactico = df_completos.select(
    "zona_geografica",
    "precio_noche"
)

df_tactico = df_tactico.withColumn(
    "precio_num",
    F.col("precio_noche").cast("double")
)


kpi_tactico = df_tactico.groupBy("zona_geografica").agg(
    F.round(F.min("precio_num"),0).alias("Precio_Minimo"),
    F.round(F.avg("precio_num"),0).alias("Precio_Promedio"),
    F.round(F.max("precio_num"),0).alias("Precio_Maximo")
)


print("[KPI TACTICO] - BANDAS DE PRECIOS POR ZONA")

kpi_tactico.show()
```

```python
import matplotlib.pyplot as plt

df_tac_pandas = (
    kpi_tactico
    .limit(15)
    .toPandas()
)


plt.figure(figsize=(12,6))

plt.plot(
    df_tac_pandas["zona_geografica"],
    df_tac_pandas["Precio_Promedio"],
    marker="o"
)

plt.xticks(rotation=45)

plt.title(
    "VISTA TACTICA: Precio Promedio por Zona Geografica"
)

plt.xlabel("Zona Geografica")
plt.ylabel("Precio Promedio CLP")

plt.grid(True)

plt.tight_layout()

plt.show()
```

## Código para el Nivel 3: Operacional

```python
import pyspark.sql.functions as F


promedio_precio = (
    df_completos
    .select(
        F.col("precio_noche").cast("double")
        .alias("precio")
    )
    .agg(
        F.avg("precio")
    )
    .collect()[0][0]
)


kpi_operacional_alertas = df_completos.filter(
    (F.col("puntuacion").cast("double") < 7) &
    (F.col("precio_noche").cast("double") > promedio_precio)
).select(
    "nombre_hotel",
    "plataforma",
    "zona_geografica",
    "precio_noche",
    "puntuacion"
)


print(
    "Total alertas:",
    kpi_operacional_alertas.count()
)


kpi_operacional_alertas.show(10)
```

```python
import matplotlib.pyplot as plt


df_op_pandas = (
    df_completos
    .select(
        F.col("precio_noche").cast("double").alias("precio"),
        F.col("puntuacion").cast("double").alias("puntuacion")
    )
    .dropna()
    .limit(500)
    .toPandas()
)


plt.figure(figsize=(10,6))


plt.scatter(
    df_op_pandas["puntuacion"],
    df_op_pandas["precio"]
)


plt.axvline(
    7,
    linestyle="--"
)

plt.axhline(
    promedio_precio,
    linestyle="--"
)


plt.title(
    "Calidad vs Precio de Alojamientos"
)

plt.xlabel("Puntuacion")

plt.ylabel("Precio CLP")


plt.grid(True)

plt.show()
```

```python
# Resumen de clusters con tus columnas reales

resumen_clusters = df_historico.groupBy("prediction") \
    .agg(
        F.count("*").alias("Cantidad_Registros"),
        
        F.round(
            F.avg(F.col("precio_raw")),
            2
        ).alias("Precio_Promedio"),
        
        F.round(
            F.avg(F.col("rating")),
            2
        ).alias("Rating_Promedio"),
        
        F.round(
            F.avg(F.col("opiniones")),
            2
        ).alias("Opiniones_Promedio")
    ) \
    .orderBy("prediction")


print("Perfil de clusters")

resumen_clusters.show()
```

# Acto 2: El Descubrimiento del Mercado (Segmentación)

```python
df_clusters_pandas = resumen_clusters.toPandas()


etiquetas = {
    0:"Segmento 0",
    1:"Segmento 1",
    2:"Segmento 2"
}


df_clusters_pandas["Segmento"] = (
    df_clusters_pandas["prediction"]
    .map(etiquetas)
)


import matplotlib.pyplot as plt


plt.figure(figsize=(10,6))


plt.bar(
    df_clusters_pandas["Segmento"],
    df_clusters_pandas["Precio_Promedio"]
)


plt.title("Precio promedio por segmento")
plt.xlabel("Cluster")
plt.ylabel("Precio promedio")


for i,v in enumerate(df_clusters_pandas["Precio_Promedio"]):
    plt.text(
        i,
        v,
        f"{v:.0f}",
        ha="center"
    )


plt.show()


print("Interpretacion:")

for _,row in df_clusters_pandas.iterrows():

    print(
        f"Cluster {int(row['prediction'])}: "
        f"{row['Cantidad_Registros']} registros | "
        f"Precio promedio {row['Precio_Promedio']} | "
        f"Rating {row['Rating_Promedio']}"
    )
```

Si observamos la tabla 1. El Cluster 0 representa el 63.9% del mercado con un precio bajo (Gama Económica). El Cluster 2 tiene pocos productos (75) pero un precio altísimo (Gama Premium).

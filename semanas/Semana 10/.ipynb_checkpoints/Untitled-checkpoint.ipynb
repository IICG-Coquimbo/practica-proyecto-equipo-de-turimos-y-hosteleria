```python
import os
import json

ruta_base = "/home/jovyan/work"

corruptos = []

for root, dirs, files in os.walk(ruta_base):
    for file in files:
        if file.endswith(".ipynb"):
            ruta_completa = os.path.join(root, file)
            try:
                with open(ruta_completa, "r", encoding="utf-8") as f:
                    json.load(f)
            except Exception as e:
                corruptos.append(ruta_completa)

print(f"Notebooks corruptos encontrados: {len(corruptos)}")
for c in corruptos:
    print(c)
```

```python
import re
import json

corruptos = [
    "/home/jovyan/work/semanas/Semana 4 Mongo DB/MongoScraper.ipynb",
    "/home/jovyan/work/semanas/Semana 5 MongoDB PySpark/Procedimiento_Spark.ipynb",
    "/home/jovyan/work/semanas/Semana 5 MongoDB PySpark/Scraping_Mongo.ipynb",
    "/home/jovyan/work/semanas/Semana 6 Integracion/Amazon.ipynb",
    "/home/jovyan/work/semanas/Semana 7 La union/ConexionMongoDBAtlas.ipynb",
    "/home/jovyan/work/semanas/Semana 9. EDA/Semana_9_EDA.ipynb",
    "/home/jovyan/work/Turismo y hoteleria/Untitled1.ipynb"
]

for ruta in corruptos:
    with open(ruta, "r", encoding="utf-8") as f:
        contenido = f.read()
    
    # Eliminar marcas de conflicto de Git, quedándose con la versión HEAD
    limpio = re.sub(r'<<<<<<< HEAD\n(.*?)=======.*?>>>>>>> [^\n]+\n', 
                    r'\1', contenido, flags=re.DOTALL)
    
    with open(ruta, "w", encoding="utf-8") as f:
        f.write(limpio)
    
    print(f"Reparado: {ruta}")

print("\nListo, recarga JupyterLab (F5) y verifica los notebooks")
```

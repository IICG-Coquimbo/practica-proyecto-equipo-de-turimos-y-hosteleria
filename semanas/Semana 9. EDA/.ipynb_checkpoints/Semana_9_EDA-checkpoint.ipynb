# PAso 1: Conexión con Mongo Atlas

```python
import os
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, DoubleType
from pymongo import MongoClient
import certifi

print("🚀 Inicializando entorno de PySpark con campos reales de Turismo...")

# 1. Creamos la sesión de Spark
spark = SparkSession.builder \
    .appName("EDA_Turismo_AngeloRojo") \
    .getOrCreate()

# 2. Conexión directa mediante PyMongo (Usa tu URI verificada del cluster)
uri = "mongodb+srv://AngeloRojo:azul2003@cluster0.aibpfux.mongodb.net/?retryWrites=true&w=majority"

print("🛰️ Conectando y extrayendo datos desde tu MongoDB Atlas...")
try:
    client = MongoClient(uri, tlsCAFile=certifi.where())
    db = client['proyecto_bigdata']
    collection = db['viajes_chile_limpios']
    
    # Extraemos todos los documentos de la colección
    documentos_raw = list(collection.find())
    
    # Limpieza rápida de tipos de datos de Python antes de pasarlos a Spark
    for doc in documentos_raw:
        if '_id' in doc:
            doc['_id'] = str(doc['_id'])
        if 'fecha_captura' in doc and doc['fecha_captura'] is not None:
            doc['fecha_captura'] = str(doc['fecha_captura'])
        # Aseguramos que el precio sea flotante puro y manejamos valores no numéricos
        if 'precio_num' in doc:
            try:
                import math
                if math.isnan(doc['precio_num']):
                    doc['precio_num'] = None
            except:
                pass

    print(f"📦 Datos recuperados con éxito. Documentos en caché: {len(documentos_raw)}")

    # 3. Definición del ESQUEMA REAL que descubrimos en las entrañas de tu BD
    schema = StructType([
        StructField("_id", StringType(), True),
        StructField("actividad", StringType(), True),
        StructField("ubicacion", StringType(), True),
        StructField("precio_texto", StringType(), True),
        StructField("precio_num", DoubleType(), True),
        StructField("fecha_captura", StringType(), True),
        StructField("plataforma", StringType(), True),
        StructField("estudiante", StringType(), True)
    ])

    # 4. Creamos el DataFrame oficial de PySpark
    df_base = spark.createDataFrame(documentos_raw, schema=schema)
    
    # 5. ESTANDARIZACIÓN OBLIGATORIA DE LA PAUTA
    # Renombramos 'ubicacion' o 'actividad' como 'identificador' y 'precio_num' como 'valor'
    df_raw = df_base.withColumnRenamed("actividad", "identificador") \
                    .withColumnRenamed("precio_num", "valor")
    
    print("\n✅ ¡DATOS CARGADOS Y ESTANDARIZADOS EN PYSPARK EXITOSAMENTE!")
    print(f"📊 Total de registros listos para el EDA de la Semana 9: {df_raw.count()}")
    
    # Mostrar una muestra con los datos reales en formato Spark
    df_raw.select("_id", "identificador", "ubicacion", "valor", "plataforma").show(5, truncate=False)

except Exception as e:
    print(f"❌ Error crítico durante la carga de datos: {e}")

```

```python
# Miramos el primer documento extraído directamente para ver sus llaves reales
if documentos_raw:
    print("🔑 Los campos reales de tus datos en MongoDB son:")
    print(documentos_raw[0].keys())
    print("\n📝 Y un ejemplo de registro completo se ve así:")
    print(documentos_raw[0])
else:
    print("No hay documentos en la lista.")
```

## Muestra los precios y moneda de los primero 10 registros

```python
print("==================================================")
print("📊 DATOS ESTANDARIZADOS Y CONTEO POR DESTINO")
print("==================================================")

# 1. Mostramos los primeros 10 registros con el formato exigido (identificador y valor)
df_raw.select("identificador", "ubicacion", "valor").show(10, truncate=False)

# 2. Contamos cuántos tours/hospedajes hay guardados por cada zona geográfica
df_raw.groupBy("ubicacion").count().orderBy("count", ascending=False).show(truncate=False)
```

## Cuento los datos extraidos por tienda

```python
print("==================================================")
print("📈 PRECIO PROMEDIO EN CLP POR DESTINO TURÍSTICO")
print("==================================================")

from pyspark.sql.functions import avg, round, col

# Agrupamos por destino y calculamos el promedio real de la columna 'valor'
df_raw.groupBy("ubicacion") \
      .agg(round(avg("valor"), 0).alias("Precio_Promedio_CLP")) \
      .orderBy("Precio_Promedio_CLP", ascending=False) \
      .show(truncate=False)
```

# Paso 2: Limpieza y Arreglo de Inconsistencias
## 1. Eliminación de Duplicados e Incompletos

```python
# Quitar registros exactamente iguales basados en el ID único de MongoDB
df_clean = df_raw.dropDuplicates(["_id"]) \
                 .filter(col("identificador").isNotNull()) \
                 .filter(col("valor") > 0)

print(f"📊 Cantidad de registros limpios: {df_clean.count()}")
```

```python
# Mostramos tus datos de turismo unificados y limpios
df_clean.select("identificador", "ubicacion", "valor", "plataforma").show(10, truncate=False)
```

```python
# Aseguramos que la columna de ubicación no tenga nulos y esté en mayúsculas
from pyspark.sql.functions import upper
df_clean = df_clean.withColumn("ubicacion", upper(col("ubicacion")))

df_clean.select("identificador", "ubicacion", "valor").show(10, truncate=False)
```

## 2. Normalización de Precios (CLP a EUR)

```python
# Convertimos el precio a numérico y normalizamos a una sola moneda (EUR)
df_clean = df_clean.withColumn("precio_num", col("valor").cast("double"))

df_clean.select("identificador", "ubicacion", "precio_num").show(5)
```

## 3. Extracción de Peso (Bioingeniería de Variables)

```python
from pyspark.sql.functions import when, col

# BIOINGENIERÍA DE VARIABLES: Categorizamos los tours según su rango de precio real
df_clean = df_clean.withColumn(
    "categoria_precio",
    when(col("precio_num") < 150000, "1-Económico")
    .when((col("precio_num") >= 150000) & (col("precio_num") < 400000), "2-Estándar")
    .otherwise("3-Premium")
)

# Mostramos la nueva columna creada para tu proyecto G5
df_clean.select("identificador", "ubicacion", "precio_num", "categoria_precio").show(15, truncate=False)
```

```python
from pyspark.sql.functions import col

print("--- COMPARATIVA DE TRANSFORMACIÓN DE DATOS (TURISMO) ---")
df_clean.select(
    col("identificador").alias("Original (Texto)"),
    col("ubicacion").alias("Destino"),
    col("precio_num").alias("Limpio (Precio CLP)"),
    col("categoria_precio").alias("Variable Creada")
).show(15, truncate=False)
```

# Paso 3: Análisis Exploratorio de Datos (EDA) con Spark

## 1. Análisis Descriptivo (Estadísticas básicas)

```python
# Resumen de variables numéricas (Precio y Peso)
df_clean.select("precio_num").describe().show()
```

## 2. Análisis de Valores Faltantes (Nulos)

```python
from pyspark.sql.functions import count, isnan, when, col

print("==================================================")
print("🔍 2. ANÁLISIS DE VALORES FALTANTES (NULOS)")
print("==================================================")

df_clean.select([count(when(isnan(c) | col(c).isNull(), c)).alias(c) for c in df_clean.columns]).show()
```

## 3. Análisis por Grupos (Marcas y Tiendas)

```python
print("==================================================")
print("📊 3. ANÁLISIS POR GRUPOS (CATEGORÍAS DE PRECIO)")
print("==================================================")

df_clean.groupBy("categoria_precio").count().orderBy("categoria_precio").show()
```

# Paso 4: Visualización

```python
import seaborn as sns
import matplotlib.pyplot as plt

print("🎨 4. GENERANDO VISUALIZACIÓN FINAL DEL EDA...")

# Convertimos a Pandas únicamente para poder usar las herramientas gráficas de Seaborn
pdf_turismo = df_clean.toPandas()

plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# Graficamos la distribución de precios por destino turístico chileno
sns.boxplot(x='ubicacion', y='precio_num', data=pdf_turismo, hue='ubicacion', palette="Set2", legend=False)

plt.title("Distribución de Precios por Destino Turístico (Semana 9)", fontsize=14, pad=15)
plt.xlabel("Destino Geográfico", fontsize=12)
plt.ylabel("Precio (CLP $)", fontsize=12)
plt.xticks(rotation=15)

plt.tight_layout()
plt.show()
```

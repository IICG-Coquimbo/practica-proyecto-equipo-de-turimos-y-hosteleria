```python
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.regression import LinearRegression
from pyspark.ml.evaluation import RegressionEvaluator

# 1. Preparación: incluimos 'prediction' (tu cluster) como característica
# Si tus columnas tienen nombres distintos a 'estrellas' o 'puntuacion', cámbialos aquí
assembler = VectorAssembler(inputCols=["estrellas", "puntuacion", "prediction"], outputCol="features")
df_model_ready = assembler.transform(df_clusters)

# 2. División (80/20)
train_data, test_data = df_model_ready.randomSplit([0.8, 0.2], seed=42)

# 3. Entrenamiento
lr = LinearRegression(featuresCol="features", labelCol="precio_noche")
model = lr.fit(train_data)

# 4. Evaluación: Esto es lo que confirma si tu regresión tiene sentido
predictions = model.transform(test_data)
evaluator = RegressionEvaluator(labelCol="precio_noche", predictionCol="prediction", metricName="rmse") # 'prediction' aquí es la predicción del modelo
# Si 'prediction' de la columna del cluster te causa conflicto, usa:
# evaluator = RegressionEvaluator(labelCol="precio_noche", predictionCol="prediction", metricName="rmse")

rmse = evaluator.evaluate(predictions)

print(f"--- Resultados de Regresión ---")
print(f"Intercepto: {model.intercept}")
print(f"Coeficientes: {model.coefficients}")
print(f"Error RMSE (mientras más bajo, mejor): {rmse}")
predictions.select("precio_noche", "prediction", "features").show(10)
```
